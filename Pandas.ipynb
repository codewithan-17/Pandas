```python
# String
import numpy as pd
import pandas as pd
country = ['India', 'Pakistan', 'USA', 'Nepal', 'SriLanka']
pd.Series(country)
```

```python
# Integer
runs = [13,24,56,78,100]
pd.Series(runs)
```

```python
# custom index
marks = [67,88,90,100,99]
subjects = ['eng','hindi','math','sci','gk']
pd.Series(marks,index=subjects)
```

```python
# setting a name
pd.Series(marks,index=subjects,name="Anamika ke marks")
```
